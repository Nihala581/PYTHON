# **Data Structures – Dictionary, Dictionary Methods, Set, Set Methods, and Frozenset** #


## **What is a Data Structure?** ##
    
A Data Structure is a way of organizing, storing, and managing data so it can be accessed and modified efficiently.

**Python provides several built-in data structures:**
    
List

Tuple

Dictionary

Set

Frozenset


## **Part 1: Dictionary** ##


**What is a Dictionary?** 

A Dictionary is an unordered , mutable collection of key-value pairs.
    
**Stores data as key : value**

Keys must be unique

Values can be duplicated

Mutable (can be modified)

Fast lookup using keys


## **Access Dictionary Values** ##

```python
student = {
    "name":"john",
    "age":20
}
print(student["name"])
```

## **Dictionary Methode** ##


**1. get()**

Returns the value of a key.


```python
student = {"name":"john","age":36}
print(student.get("name"))
```

**2. Keys()**

Returns all keys.


```python
student = {"name":"john","age":36}
print(student.keys())
```

**3. values()**

Returns all values.


```python
student = {"name":"john","age":36}
print(student.values())
```

**4. items()**

Returns all key-value pairs.


```python
student = {"name":"john","age":36}
print(student.items())
```

**5. update()**

Updates the dictionary.


```python
# Methode
student = {"name":"john","age":36}
student.update({"course":"data science"})
print(student)
```

```python
# with Methode
student = {"name":"john","age":36}
student.update({"age":20})
print(student)
```

**6. pop()**

Removes a specified key.


```python
student = {"name":"john","age":36}
student.pop("name")
print(student)
```

**7. popitem()**

Removes the last inserted item.


```python
student = {"name":"john","age":36}
student.popitem()
print(student)
```

**8. clear()**

Removes all items.


```python
student = {"name":"john","age":36}
student.clear()
print(student)
```

**9. copy()**

Returns a copy of the dictionary.


```python
student = {"name":"john","age":36}
new_student = student.copy()
print(new_student)
```

## **Part 2: Set** ##

**What is a Set?**

A Set is an unordered collection of unique elements.
    
**Features**

Unordered

Mutable

No duplicate values

Cannot access items using indexing
