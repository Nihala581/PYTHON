## **Part D: Coding Problems** ##

**D1. Shopping List**

```python
shopping = ["milk", "bread", "eggs"]

shopping.append("butter")
shopping.insert(1, "rice")
shopping.extend(["salt", "sugar"])
shopping.remove("bread")

print(shopping)
print(shopping.index("eggs"))

shopping.sort()
print(shopping)
```

**D2. Student Marks**

```python
marks = [72, 85, 90, 65, 88, 76, 95, 58]

marks.sort(reverse=True)

print(marks)
print("Top 3:", marks[:3])
print("Bottom 3:", marks[-3:])
print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
```

**D3. Monthly Sales Report**

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 9000, 18000, 21000, 17000]

total = sum(sales)
average = total / len(sales)

highest = max(sales)
index = sales.index(highest)
best_month = months[index]

count = 0
for sale in sales:
    if sale > average:
        count += 1

print("Total:", total)
print("Average:", round(average, 2))
print("Best month:", best_month, "(", highest, ")")
print("Months above average:", count)
```

**D5. Remove Duplicates (keep the order)**

```python
data = [3, 1, 3, 2, 1, 5, 2]

unique = []

for item in data:
    if item not in unique:
        unique.append(item)

print(unique)
```

**D6. Even and Odd Separator**

```python
numbers = [12, 7, 8, 3, 20, 15, 6, 9]

even = []
odd = []

for n in numbers:
    if n % 2 == 0:
        even.append(n)
    else:
        odd.append(n)

even.sort()
odd.sort()

print("Even:", even)
print("Odd:", odd)
```

**D7. Palindrome List**

```python
a = [1, 2, 3, 2, 1]
b = ["a", "b", "c"]

if a == a[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")

if b == b[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```
