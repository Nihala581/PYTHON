## **Python Dictionary Methods — Assignment** ##

**Set A — Warm-up**

```python
create_book = {"title": "Python basics" , "price": 300}
create_book.update({"author": "Ravi"})
print(create_book)
```

```python
Start_with_book = {"title": "Python Basics" , "price": 300}
Start_with_book.update({"price": 250})
Start_with_book.update({"pages": 180})
print(Start_with_book)
```

```python
start_with_book = {"title": "Python Basics", "price": 250, "pages": 180}
start_with_book.pop("pages")
print(start_with_book)
```

```python
fruits = {"apple": 50, "banana": 30, "mango": 20}
fruits.popitem()
print(fruits)
```

```python
fruits = {"apple": 50, "banana": 30, "mango": 20}
fruits_copy = fruits.copy()
fruits.clear()
print("original:",fruits)
print("copy:",fruits_copy)
```

```python
d = {"a": 1, "b": 2, "c": 3}
d.pop("b")
d.popitem()
print(d)
```

**Set B — Apply It**

```python
stock = {"apple": 50, "banana": 30, "mango": 20}
stock.update({"apple": 45})
stock.update({"orange": 25})
print(stock)
```

```python
stock = {"apple": 50, "banana": 30, "mango": 20}
key = input("enter a key to delete:")
removed = stock.pop(key, "key not found")

print(removed)
print(stock)
```

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 3}
dict1.update(dict2)
print(dict1)
```

```python
a = {"x": 1}
b = a
b["x"] = 99
print(a)
```

```python
marks = {"Maths": 80, "Science": 75, "English": 90}
while marks: 
    item = marks.popitem()
    print(item)
```

```python
cart = {"apple": 2, "banana": 3, "milk": 1}

backup = cart.copy()

cart.clear()

print("Cart:", cart)
print("Backup:", backup)
print("Cart length:", len(cart))
print("Backup length:", len(backup))
```
