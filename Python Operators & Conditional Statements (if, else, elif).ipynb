## **Python Operators & Conditional Statements (if, else, elif)** ##


## **Part 1: Python Operators** ##

**What is an Operator?**

An Operator is a symbol that performs operations on variables and values.


**Q - Write a Python program to add two numbers.**


```python
a = 10
b = 45
print(a+b)
```

**Types of Operators in Python**

Python provides the following commonly used operators:
    
 - Arithmetic Operators
 - Comparison (Relational) Operators
 - Assignment Operators
 - Logical Operators


## **1. Arithmetic Operators** ##


**Q - Write a Python program to perform all arithmetic operations on two numbers.**


```python
A = 20
B = 60

print("addition:",A+B)
print("substraction:",A-B)
print("multiplication:",A*B)
print("division:",A/B)
print("floor divition:",A//B)
print("modulus:",A%B)
print("exponent:",A**B)
```

## **2. Comparison (Relational) Operators** ##

**Q - Write a Python program to compare two numbers using comparison operators.**


```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

## **3. Assignment Operators** ##


**Q - Write a Python program to demonstrate assignment operators.**


```python
x = 10

x += 5
print(x)

x -= 2
print(x)

x *= 3
print(x)

x /= 2
print(x)
```

## **4. Logical Operators** ##


**Q- Write a Python program to demonstrate logical operators.**


```python
age = 25
salary = 30000

print(age > 18 and salary > 20000)
print(age > 30 or salary > 20000)
print(not(age > 30))
```

## **Part 2: Conditional Statements** ##
**What are Conditional Statements?**

Conditional statements allow Python to make decisions based on conditions.
Python provides:
- if
- if...else
- if...elif...else


## **1. if Statement** ##


**Q - Write a Python program to check whether a person is eligible to vote.**


```python
age = 20

if age >= 18:
    print("Eligible to Vote")

```

**Q - Write a Python program to check whether a student has passed.**


```python
marks = 85

if marks >= 50:
    print("Pass")

```

## **2. if...else Statement** ##


**Q - Write a Python program to check whether a person is eligible to vote.**


```python
age = 15

if age >= 18:
    print("Eligible")
else:
    print("Not Eligible")
```

**Q - Write a Python program to check whether a number is even or odd.**


```python
number = 8

if number % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

**Q - Write a Python program to find the greater of two numbers.**


```python
a = 15
b = 25

if a > b:
    print("A is Greater")
else:
    print("B is Greater")
```

## **3. if...elif...else Statement** ##


**Q - Write a Python program to calculate the grade of a student.**


```python
marks = 72

if marks >= 90:
    print("Grade A+")

elif marks >= 75:
    print("Grade A")

elif marks >= 60:
    print("Grade B")

elif marks >= 50:
    print("Grade C")

else:
    print("Fail")
```

**Q - Write a Python program to display the weather condition based on temperature.**


```python
temperature = 38

if temperature > 40:
    print("Very Hot")

elif temperature >= 30:
    print("Hot")

elif temperature >= 20:
    print("Pleasant")

else:
    print("Cold")
```

**Q - Write a Python program to calculate the discount based on the purchase amount.**


```python
amount = 4500

if amount >= 10000:
    print("30% Discount")

elif amount >= 5000:
    print("20% Discount")

elif amount >= 2000:
    print("10% Discount")

else:
    print("No Discount")
```

**Q - Write a Python program to check loan eligibility.**


```python
age = 25
salary = 50000

if age > 18 and salary > 30000:
    print("Loan Approved")
else:
    print("Loan Rejected")

```

**Q - Write a Python program to create a simple login system.**


```python
username = "admin"
password = "1234"

if username == "admin" and password == "1234":
    print("Login Successful")
else:
    print("Invalid Credentials")

```
