### **Section B: Coding Problems** ###

**B1. Name Formatter**

Take a full name as input, for example " aDiTHYa p t ". Clean it (remove extra spaces) and print it in
Title Case.
Expected: Adithya P T

```python
name = " aDiTHYa p t "
name = name.title()
name = name.strip()
print(name)
```

**B2. Email Validator (basic)**

Take an email as input and print Valid only if all of these are true:
• It contains @
• It ends with .com
• It does not start with @
Try: student@gmail.com, @gmail.com and student@gmail.in

```python
email = input("enter email: ")

if "@" in email and email.endswith(".com") and not email.startswith("@"):
    print("Valid")
else:
    print("Invalid")
```

**B3. Username Extractor**

From an email like student@gmail.com, print the username (student) and the domain (gmail.com).
Hint: find() and slicing.

```python
email = input("enter email:")

at_position = email.find("@")

username = email[:at_position]
domain = email[at_position + 1:]

print("username:", username)
print("domain:", domain)
```

**B4. Palindrome Checker**

Take a word and check whether it is a palindrome, ignoring case.
Madam → Palindrome Python → Not a Palindrome

```python
word = input("enter a word:")

word = word.lower()

if word == word[::-1]:
    print("Palindrom")
else:
    print("Not a Palindrom")
```

**B5. Mask the Phone Number**

For a 10-digit number like "9876543210", print it as 98******10.
Hint: slicing + string multiplication.

```python
phone = input("enter phone number:")

masked = phone[:2] + "*" * 6 + phone[-2:]

print(masked)
```

**B6. Initials Generator**

Given "Adithya Kumar Nair", print the initials as A.K.N
Hint: split(), indexing, join().

```python
name = input("enter full name:")

words = name.split()

initials = ".".join(word[0].upper() for word in words)

print(initials)
```

**B7. Swap First and Last Character**

Given "python", print nythop. Use slicing only.

```python
word = "python"

result = word[-1] + word[1:-1] + word[0]

print(result)
```

**B8. Data Cleaning: Course List**

raw = " python ,SQL, power bi ,EXCEL, gcp "
Convert this into a clean list: ['Python', 'Sql', 'Power Bi', 'Excel', 'Gcp']
Then join it with " | " and print the result.

```python
raw = "python ,SQL ,power bi ,EXCEL ,gcp"

courses = [course.strip().title() for course in raw.split(",")]

result = " | ".join(courses)

print(result)
```

### **Section C: Mini Task - Student ID Card** ###

```python
data = " rahul kumar,data analytics,rahul.kumar@gmail.com,9876543210 "

# Remove extra spaces
data = data.strip()

# Split the data
details = data.split(",")

# Get each field
name = details[0].strip().title()
course = details[1].strip().title()
email = details[2].strip()
phone = details[3].strip()

# Username
at_position = email.find("@")
username = email[:at_position]

# Mask phone number
masked_phone = phone[:2] + "*" * 6 + phone[-2:]

# Initials
name_parts = name.split()
initials = ".".join(part[0].upper() for part in name_parts)

# Course code
course_code = course[:3].upper() + "-2026"

# Print Student ID Card
print("========== STUDENT ID CARD ==========")
print("Name      :", name)
print("Course    :", course)
print("Email     :", email)
print("Username  :", username)
print("Phone     :", masked_phone)
print("Initials   :", initials)
print("Course Code:", course_code)
print("=====================================")
```
