```python
num = int(input("enter a number: "))

if num % 3 == 0 and num % 5 == 0:
    print("FizzBuzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
else:
    print(num)
```

```python
seconds = int(input("enter total seconds: "))
hours = seconds // 3600
remaining = seconds % 3600
minutes = remaining // 60
seconds = remaining % 60
print(hours, "hours", minutes, "minutes", seconds,"seconds")
```

```python
weight = float(input("enter weight in kg: "))
height = float(input("enter height in meters: "))
bmi = weight / (height ** 2)
print("BMI:",bmi)
if bmi < 18.5:
    print("underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("overweight")
else:
    print("obese")
```

```python
num = int(input("Enter a three-digit number: "))

first = num // 100
last = num % 10

if first == last:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
cp = float(input("Enter cost price: "))
sp = float(input("Enter selling price: "))

if sp > cp:
    profit = sp - cp
    percentage = (profit / cp) * 100
    
    print("Profit:", profit)
    print("Profit Percentage:", percentage, "%")

elif sp < cp:
    loss = cp - sp
    percentage = (loss / cp) * 100
    
    print("Loss:", loss)
    print("Loss Percentage:", percentage, "%")

else:
    print("No Profit, No Loss")
```

```python
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

D = b ** 2 - 4 * a * c

print("Discriminant:", D)

if D > 0:
    print("Roots are real and different")
elif D == 0:
    print("Roots are real and equal")
else:
    print("Roots are imaginary")
```

```python
distance = float(input("Enter distance in km: "))
hour = int(input("Enter hour (0 to 23): "))

if distance <= 1:
    fare = 60
else:
    fare = 60 + (distance - 1) * 15

if hour >= 22 or hour < 5:
    fare = fare * 1.25

print("Final fare:", fare)
```

```python
x = float(input("Enter x: "))
y = float(input("Enter y: "))

if x == 0 and y == 0:
    print("Origin")
elif y == 0:
    print("On the x-axis")
elif x == 0:
    print("On the y-axis")
elif x > 0 and y > 0:
    print("Quadrant 1")
elif x < 0 and y > 0:
    print("Quadrant 2")
elif x < 0 and y < 0:
    print("Quadrant 3")
else:
    print("Quadrant 4")
```

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))

a = a + b
b = a - b
a = a - b

print("After swapping:")
print("a =", a)
print("b =", b)

if a > b:
    print("a is greater")
elif b > a:
    print("b is greater")
else:
    print("Both are equal")
```
