```python
import pandas as pd
data={
    "Name":["Arun","Rahul","Anu","John","Meera"],
    "Maths":[85,70,95,60,80],
    "Science":[90,75,88,65,85],
    "English":[80,85,90,70,95]}
df=pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Name"],df["Maths"])
plt.title("exam reault")
plt.xlabel("Name")
plt.ylabel("Maths")
plt.show
```

```python
plt.plot(df["Name"],df["English"],marker = "o")
plt.title("exam result")
plt.xlabel("Name")
plt.ylabel("English")
plt.show
```

```python
plt.pie(df["Maths"],labels=df["Name"],autopct = "%1.2f%%")
```

```python
plt.hist(df["Maths"],bins = 9, edgecolor = "Black")
```
