## **Part A: While + Break (stop when you find what you need)** ##


```python
sales = [1200, 1500, 900, 2500, 1800]
i = 0
n = len(sales)
print(n)
while i <= n:
    
    if sales[i]>2000:
        print("highest sales day-day",i+1)
        print("highest sales=" , sales[i])
        break
    i+=1
    
```

```python
marks = [78, 65, 90, 28, 55]
i = 0
n = len(marks)
print(n)
while i <= n:

    if marks[i]<35:
        print("failed student:",i+1)
        print("low marks:", marks[i])
        break
    i+=1
```

```python
sales = [1200, 1500, 900, 2000, 1800]
i = 0
total = 0
n = len(sales)
print(n)
while i<= n:
    total += sales[i]
    if total>5000:
        print("number of day:", i+1)
        print("total sales:",total)
        break
    i+=1
```

## **Part B: While + Continue (skip bad or unwanted values)** ##


```python
temperatures = [31, None, 34, None, 29]

i = 0

while i < len(temperatures):
    if temperatures[i] is None:
        i += 1
        continue
    print(temperatures[i])
    i += 1
```

```python
profit = [500, -200, 800, -100, 300]

i = 0
total = 0

while i < len(profit):
    if profit[i] < 0:
        i += 1
        continue
    total = total + profit[i]
    i += 1

print(total)
```

```python
orders = [3, 0, 5, 0, 2]

i = 0
count = 0

while i < len(orders):
    if orders[i] == 0:
        i += 1
        continue
    count += 1
    i += 1

print(count)
```

## **Part C: while + break + continue together** ##


```python
data = [100, None, 250, None, 300, -1, 400]

i = 0
total = 0

while i < len(data):
    if data[i] == -1:
        break
    if data[i] is None:
        i += 1
        continue
    total = total + data[i]
    i += 1

print(total)
```

```python
sales = [1200, 0, 1500, 0, 2500, 1800]

i = 0

while i < len(sales):
    if sales[i] == 0:
        i += 1
        continue
    if sales[i] > 2000:
        print("Target crossed")
        break
    print(sales[i])
    i += 1
```

```python
total = 0

while True:
    amount = float(input("Enter sale amount: "))

    if amount < 0:
        print("Invalid amount")
        continue

    if amount == 0:
        break

    total = total + amount

print("Total sales:", total)
```

## **Part D: while with else + break** ##


```python
ids = [101, 102, 103, 104]

search_id = int(input("Enter customer ID: "))

i = 0

while i < len(ids):
    if ids[i] == search_id:
        print("Position:", i + 1)
        break
    i += 1
else:
    print("Customer not found")
```
