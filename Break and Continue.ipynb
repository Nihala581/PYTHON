## **Break Statement** ##


**Q - Write a Python program to print numbers from 1 to 10, but stop when the number becomes 6.**


```python
i = 1

while i <= 10:

    if i == 6:
        break

    print(i)
    i += 1

```

**Q - Write a Python program to stop the loop when the number becomes 4.**


```python
i = 1

while i <= 10:

    if i == 4:
        break

    print(i)
    i += 1
```

## **Continue Statement** ##


**Q - Write a Python program to print numbers from 1 to 5 except 3.**


```python
i = 1

while i <= 5:

    if i == 3:
        i += 1
        continue

    print(i)
    i += 1
```

**Q - Write a Python program to print only odd numbers from 1 to 10.**


```python
i = 1

while i <= 10:

    if i % 2 == 0:
        i += 1
        continue

    print(i)
    i += 1
```
