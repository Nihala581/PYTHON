```python
sales = [1200, 1500, 900, 2000, 1800]
i = 0
while i < len(sales):
    print(sales[i])
    i = i+1
```

```python
sales = [1200, 1500, 900, 2000, 1800]
i = 0
total = 0

while i < len(sales):
    total = total + sales[i]
    i = i+1
print(total)
    
    
```

```python
marks = [45, 78, 62, 90, 55]
i = 0
total = 0
while i <len(marks):
    total = total +marks[i]
    i = i+1
average = total/len(marks)
    
print(average)

```

```python
marks = [45, 78, 62, 90, 55]

i = 0
count = 0

while i < len(marks):
    if marks[i] >= 50:
        count = count + 1
    i = i + 1

print(count)
```

```python
quantity = int(input("Enter quantity: "))

i = 1

while i <= quantity:
    print(i * 25)
    i = i + 1
```

```python
temperatures = [31, 34, 29, 36, 33]

i = 1
highest = temperatures[0]

while i < len(temperatures):
    if temperatures[i] > highest:
        highest = temperatures[i]
    i = i + 1

print(highest)
```

## **Part B: While With Else** ##

```python
i = 1

while i <= 3:
    print(i)
    i = i + 1
else:
    print("Report Generated")
```
