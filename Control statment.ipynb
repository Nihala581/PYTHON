**What is a While Loop?**


**1 - Write a Python program to print numbers from 1 to 10.**


```python
i = 1
while i<=10:
    print(i)
    i+=1
```

**2 - Write a Python program to print numbers from 5 to 1.**


```python
i = 5
while i>= 1:
    print(i)
    i-=1
```

**3 - Write a Python program to find the sum of the first 10 natural numbers.**


```python
n = 10
i = 0
total = 0

while i <= n:
    total +=i
    i +=1
else :
    print("loop finished")
print("total sum:", total)
    
```

 **While Loop with else** 


The else block runs after the while loop finishes normally.
Syntax:
while condition:
    statements
else:
    statements


```python
i = 1

while i <= 3:
    print(i)
    i += 1

else:
    print("Loop Finished")

```

**Question: A shop's sales for 5 days are stored in a list. Write a Python program to find the total and average sales.**

sales = [1200, 1500, 900, 2000, 1800]

```python
sales = [1200, 1500, 900, 2000, 1800]
total = 0
i = 0
n = len(sales)
print(n)
while i < n:
    total = total + sales[i]
    i = i + 1
average = total/len(sales)
print("total sales:", total)
print("average sales:", average)
```

## **break Statement** ##


**Q- 1 - Write a Python program to print numbers from 1 to 10, but stop when the number becomes 6.**


```python
i = 1

while i <= 10:

    if i == 6:
        break

    print(i)
    i += 1
```
