## **Part 2: Set** ##


```python
fruits = {"Apple", "Orange", "Mango"}

print(fruits)

```

**Set Methods**


**1. add()**

Adds one element.


```python
fruits = {"Apple", "Orange", "Mango"}
fruits.add("pinapple")
print(fruits)
```

**2. update()**

Adds multiple items.


```python
fruits = {"apple","lemon","watermelon"}
fruits.update(["orange","grape"])
print(fruits)
```

**3. remove()**

Removes an item.


```python
fruits = {"apple","lemon","watermelon"}
fruits.remove("lemon")
print(fruits)
```

**4. discard()**

Removes an item if present.


```python
fruits = {"apple","lemon","watermelon"}
fruits.discard("lemon")
print(fruits)
```

**5. pop()**

Removes a random item.


```python
fruits = {"apple","lemon","watermelon"}
fruits.pop()
print(fruits)
```

**6. clear()**

Removes all elements.


```python
fruits = {"apple","lemon","watermelon"}
fruits.clear()
print(fruits)
```

**7. copy()**

Returns a copy of the set.


```python
fruits = {"apple","lemon","watermelon"}
new_fruits = fruits.copy()
print(new_fruits)
```

**8. union()**

Combines two sets.


```python
A = {1,2,3,4}
B = {10,9,8,7}
print(A.union(B))
```

**9. intersection()**

Returns common elements.


```python
A = {12,13,14,35}
B = {13,45,35,67}
print(A.intersection(B))
```

**10. difference()**

Returns elements present only in the first set.


```python
A = {4,8,12}
B = {2,4,6}
print(A.difference(B))
```

**11. symmetric_difference()**

Returns elements that are in either set but not in both.


```python
A = {4,8,12}
B = {2,4,6}
print(A.symmetric_difference(B))
```

## **Part 3: Frozenset** ##


**What is a Frozenset?**

A Frozenset is an immutable version of a set. Once created, its elements cannot be added, removed, or changed.
    
**Features**

Unordered

Immutable

Unique elements only

Supports set operations like union(), intersection(), and difference()


```python
numbers = frozenset([1,2,3,4,5,6,7])
print(numbers)
```
