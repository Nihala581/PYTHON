## **What is numpy** ##

numpy stands for numerical python.

```python
sales = [15000,30000,40000,20000]
new_sales = []

for i in sales:
    new = i + (i*10/100)
    new_sales.append(new)

print(new_sales)
```

```python
import numpy as np

sales = np.array ([15000,30000,40000,20000])
new_sales = sales*1.10

print(new_sales)
```

```python
numbers = [10,20,30]
print(numbers*2)
```

```python
numbers = np.array([10,20,30])
print(numbers*2)
```

```python
sales = np.array([12000,13000,15000,32000])

print(np.mean(sales))
print(np.max(sales))
print(np.min(sales))
print(np.sum(sales))
print(np.std(sales))
print(np.median(sales))
print(np.var(sales))
print(np.sort(sales))
print(np.argmax(sales))
print(np.argmin(sales))
print(np.abs(sales))
print(np.round(sales))
print(np.cumsum(sales))
print(np.sqrt(sales))
print(np.power(sales,2))
print(np.floor(sales))
print(np.ceil(sales))
```

```python
numbers = [12,13,14,15]
print(np.sort(numbers)[:-1])
```

```python
old = np.array([20,40,60])
new = np.array([10,20,30])

percentage_change = (new-old)/old*100
print(percentage_change)

```
